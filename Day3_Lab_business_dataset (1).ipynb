```python
# ==============================================================================
# CELL 0: ENVIRONMENT SETUP & DATASET INGESTION (saas_churn_data.csv)
# Run this cell first to generate the portfolio-grade dataset on disk.
# ==============================================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Configure visualization standards
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["font.size"] = 10
plt.rcParams["axes.titlesize"] = 12
plt.rcParams["axes.labelsize"] = 11

np.random.seed(42)
N = 1200

# Generate realistic enterprise SaaS customer lifecycle data
tenure = np.random.exponential(scale=18, size=N).clip(1, 72).astype(int)
contract = np.random.choice(["Month-to-Month", "One-Year", "Two-Year"], size=N, p=[0.55, 0.25, 0.20])
support_tickets = np.random.poisson(lam=1.8, size=N) + (contract == "Month-to-Month") * np.random.poisson(lam=1.2, size=N)
monthly_charges = np.random.normal(loc=75.0, scale=25.0, size=N).clip(20.0, 150.0)
tech_support = np.random.choice(["Yes", "No", "No internet"], size=N, p=[0.40, 0.45, 0.15])
payment_method = np.random.choice(["Electronic Check", "Bank Transfer", "Credit Card"], size=N, p=[0.40, 0.30, 0.30])

# Complex churn probability generator
churn_logits = (
    - 2.0
    + 0.03 * monthly_charges
    - 0.05 * tenure
    + 0.45 * support_tickets
    + 1.2 * (contract == "Month-to-Month")
    - 1.0 * (contract == "Two-Year")
    + 0.8 * (payment_method == "Electronic Check")
)
churn_prob = 1 / (1 + np.exp(-churn_logits))
churn = np.where(np.random.rand(N) < churn_prob, "Yes", "No")

# Total spend with realistic noise and missing values
total_charges = (tenure * monthly_charges) + np.random.normal(0, 50, size=N)
total_charges[np.random.choice(N, size=15, replace=False)] = np.nan

eda_df = pd.DataFrame({
    "Customer_ID": [f"CUST-{2000+i}" for i in range(N)],
    "Tenure_Months": tenure,
    "Contract_Type": contract,
    "Payment_Method": payment_method,
    "Tech_Support": tech_support,
    "Support_Tickets": support_tickets,
    "Monthly_Charges": np.round(monthly_charges, 2),
    "Total_Charges": np.round(total_charges, 2),
    "Churn": churn
})

eda_df.to_csv("saas_churn_data.csv", index=False)
print("✅ Synthetic dataset 'saas_churn_data.csv' successfully created!")
```

```python
# ==============================================================================
# TASK 3: MISSING VALUE & OUTLIER TREATMENT (EDA CONTEXT)
# ==============================================================================

# ------------------------------------------------------------------------------
# 3.1 Missing Value Treatment in Total_Charges
# Context: Total_Charges has 15 missing values.
# Task: Impute missing Total_Charges using the formula (Tenure_Months * Monthly_Charges).
#       Explain in a comment why this is superior to simple mean/median imputation.
# ------------------------------------------------------------------------------

# TODO: Impute Total_Charges using calculated tenure * monthly charges
# df['Total_Charges'] = df['Total_Charges'].fillna(df['Tenure_Months'] * df['Monthly_Charges'])
df['Total_Charges'] = df['Total_Charges'].fillna(
    df['Tenure_Months'] * df['Monthly_Charges']
    )
print(f"Missing values remaining in Total_Charges: {df['Total_Charges'].isna().sum()}")


# ------------------------------------------------------------------------------
# 3.2 Outlier Investigation: Support Tickets
# Task: Identify accounts where Support_Tickets > (Q3 + 1.5 * IQR).
#       Determine the churn rate among these high-ticket outliers.
# ------------------------------------------------------------------------------
# TODO: Calculate IQR fences for Support_Tickets
# q1_tix = ...
# q3_tix = ...
# iqr_tix = ...
# upper_fence = ...

# outlier_accounts = df[df['Support_Tickets'] > upper_fence]
# print(f"Identified {len(outlier_accounts)} ticket-volume outliers.")
# print(f"Churn Rate in Outlier Accounts: {(outlier_accounts['Churn'] == 'Yes').mean() * 100:.2f}%")
q1_tix = df['Support_Tickets'].quantile(0.25)

q3_tix = df['Support_Tickets'].quantile(0.75)

iqr_tix = q3_tix - q1_tix

upper_fence = q3_tix + (1.5 * iqr_tix)

# Identify accounts with Support_Tickets greater than the upper fence
outlier_accounts = df[df['Support_Tickets'] > upper_fence]

print(f"Identified {len(outlier_accounts)} ticket-volume outliers.")

print(
    f"Churn Rate in Outlier Accounts: "
        f"{(outlier_accounts['Churn'] == 'Yes').mean() * 100:.2f}%"
        )
```

```python
# ==============================================================================
# TASK 4: HYPOTHESIS-DRIVEN FEATURE ENGINEERING
# ==============================================================================
# Task: Engineer at least TWO new features that capture behavioral or financial risk.
#
# Feature Idea 1: `Ticket_Velocity` = Support_Tickets / (Tenure_Months + 1)
#   (Hypothesis: High support requests in early tenure signals severe onboarding friction)
#
# Feature Idea 2: `High_Risk_Flag` = 1 if (Contract == 'Month-to-Month' and Tech_Support == 'No') else 0
#   (Hypothesis: Unassisted flexible contracts have compounded churn risk)
# ------------------------------------------------------------------------------

# TODO: Create Feature 1
df['Ticket_Velocity'] = df['Support_Tickets'] / (df['Tenure_Months'] + 1)

# TODO: Create Feature 2
df['High_Risk_Flag'] = (
    (df['Contract_Type'] == 'Month-to-Month') & (df['Tech_Support'] == 'No')
    ).astype(int)

# Validate predictive utility: Compare churn rates across your engineered features
print("--- Churn Rate Comparison for Engineered Features ---")

print("\n1) Churn rate by High_Risk_Flag (%):")
display(df.groupby('High_Risk_Flag')['Churn'].value_counts(normalize=True).unstack() * 100)

print("\n2) Ticket_Velocity distribution by Churn status:")
display(df.groupby('Churn')['Ticket_Velocity'].describe())
```

📊 Strategic Insight 1: Early Tenure is the Danger Zone for Churn

Claim: Customers who churn have substantially shorter tenure than customers who stay, indicating churn risk is concentrated in the early customer lifecycle rather than being evenly distributed.

Evidence: Median tenure for churned customers is 10 months (IQR: 4–20) vs. 18 months (IQR: 7–38) for retained customers — a difference of 8 months, with churned customers' entire IQR range sitting below the retained group's median. (Source: Task 1.1 tenure_stats)

Business Action: Launch a structured 90-day onboarding program (proactive check-ins at day 7, 30, 60, 90) specifically targeting customers in their first quarter, since this is where the bulk of churn risk is concentrated.


📊 Strategic Insight 2: Month-to-Month Contracts Without Tech Support Are a Compounded Risk Segment

Claim: Customers on flexible, unassisted contracts churn at a dramatically higher rate than any other segment, confirming that contract flexibility and lack of support compound rather than act independently.

Evidence: The High_Risk_Flag segment (Month-to-Month + No Tech Support) shows a churn rate of 86.14%, compared to 63.99% for all other customers — roughly 1.35x higher, and well above the already-high 69.58% baseline churn rate in this dataset. (Source: Task 4, High_Risk_Flag validation; cross-check against Task 1.2 crosstab)

Business Action: Bundle a free or discounted tech-support tier automatically into month-to-month plans for the first 3–6 months, converting the highest-risk segment into a supported one before it becomes a retention statistic.


📊 Strategic Insight 3: Rising Support Ticket Volume Signals Imminent Churn
Claim: Customers who file an unusually high number of support tickets (statistical outliers) churn at a meaningfully higher rate than the overall customer base, making ticket volume a leading indicator rather than a lagging one.

Evidence: 14 accounts (out of 1200 total) were flagged as ticket-volume outliers (> Q3 + 1.5×IQR = 6 tickets). Churn rate among these outliers is 100.00%, versus 69.58% for the full customer base — every single outlier account churned. (Source: Task 3.2)

Business Action: Build an automated alert that flags any account crossing the outlier ticket threshold and routes it to a senior retention specialist for a proactive save-call within 48 hours, rather than waiting for a cancellation request.


📊 Strategic Insight 4: Onboarding-Phase Ticket Velocity Predicts Churn Better Than Raw Ticket Count

Claim: It isn't the number of support tickets alone that predicts churn — it's how quickly those tickets accumulate relative to how new the customer is, confirming an onboarding-friction hypothesis rather than a generic "unhappy customer" hypothesis.

Evidence: Mean Ticket_Velocity for churned customers is 0.415 (median 0.200), compared to 0.168 (median 0.065) for retained customers — churners show roughly 2.5x higher mean ticket velocity. (Source: Task 4, Ticket_Velocity validation)
Business Action: Assign a dedicated onboarding specialist to any account that files more than 2 tickets within their first 60 days (Ticket_Velocity > 0.2, matching churners' median), treating high early ticket velocity as a distinct escalation path separate from standard support.


📊 Strategic Insight 5: Billing Amount Has a Moderate Relationship With Churn, But Tenure and Ticket Volume Matter More

Claim: Monthly and total charges are moderately correlated with each other, and monthly charges show a modest positive link to churn — but tenure and support-ticket volume are stronger churn drivers than price alone.

Evidence: Pearson correlation between Monthly_Charges and Total_Charges is 0.297, and Spearman is 0.266 (moderate, weaker than a pure product relationship would suggest, due to injected noise). In the Task 2.1 heatmap, Churn_Flag's correlation with Monthly_Charges is 0.248, close to but slightly weaker than its correlation with Support_Tickets (0.297) and notably weaker than the negative correlation with Tenure_Months (-0.256). (Source: Task 1.3 and Task 2.1)

Business Action: Deprioritize blanket price-based retention offers (e.g., across-the-board discounts) in favor of targeted service-quality interventions (onboarding support, ticket-velocity monitoring) which the data shows are comparably or more predictive churn levers than price.
